```python
# A4.1
anzahl_stuetzen = 0
for stab in tragwerk['staebe']:
    if stab['typ'] == "Stütze":
        anzahl_stuetzen = anzahl_stuetzen + 1
print(f"Anzahl Stützen: {anzahl_stuetzen}")  # 6
```

```python
# A4.2
laenge_traeger = 0.0
for stab in tragwerk['staebe']:
    if stab['typ'] == "Träger":
        laenge_traeger = laenge_traeger + stab['laenge']
print(f"Gesamtlänge Träger: {laenge_traeger:.1f} m")  # 48.0
```

```python
# A4.3
anzahl_s = anzahl_t = anzahl_q = 0
laenge_s = laenge_t = laenge_q = 0.0
for stab in tragwerk['staebe']:
    if stab['typ'] == "Stütze":
        anzahl_s += 1; laenge_s += stab['laenge']
    elif stab['typ'] == "Träger":
        anzahl_t += 1; laenge_t += stab['laenge']
    elif stab['typ'] == "Querträger":
        anzahl_q += 1; laenge_q += stab['laenge']
print(f"Stütze:      {anzahl_s}x  {laenge_s:.1f} m")
print(f"Träger:      {anzahl_t}x  {laenge_t:.1f} m")
print(f"Querträger:  {anzahl_q}x  {laenge_q:.1f} m")
```

```python
# A5.1 + A5.2
def get_staebe_by_typ(modell, typ):
    return [s for s in modell['staebe'] if s['typ'] == typ]

def berechne_gesamtlaenge(staebe):
    gesamt = 0.0
    for stab in staebe:
        gesamt = gesamt + stab['laenge']
    return gesamt

def filter_nach_laenge(staebe, min_laenge):
    ergebnis = []
    for stab in staebe:
        if stab['laenge'] >= min_laenge:
            ergebnis.append(stab)
    return ergebnis

print(berechne_gesamtlaenge(get_staebe_by_typ(tragwerk, "Träger")))  # 48.0
print(len(filter_nach_laenge(tragwerk['staebe'], 9.0)))              # 4
```

```python
# A5.3 – Freistil: Stahl in Feld 2
def filter_nach_profil(staebe, profil):
    return [s for s in staebe if s['profil'] == profil]

feld2 = filter_nach_profil(tragwerk['staebe'], "HEB 340")
print(f"Feld 2 (HEB 340): {berechne_gesamtlaenge(feld2):.1f} m")  # 28.0
```
