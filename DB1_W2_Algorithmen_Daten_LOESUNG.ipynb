# Teil 1: Algorithmen

Ein Algorithmus ist eine **eindeutige Handlungsvorschrift** zur Lösung eines Problems, bestehend aus endlich vielen, genau definierten Einzelschritten.

Das Grundprinzip ist immer gleich:

```
Eingabe → Verarbeitung nach Regeln → Ausgabe
```

Das kennt ihr schon von der von-Neumann-Architektur (Woche 1) und von eurem ersten Python-Programm.

### Beispiel: Algorithmus in Worten vs. Python

**Problem:** Prüfe, ob eine Betonplatte die Mindestdicke von 20 cm einhält.

**Algorithmus in Worten:**
1. Nimm die Dicke der Platte entgegen (Eingabe)
2. Vergleiche die Dicke mit 20 cm (Verarbeitung)
3. Wenn Dicke >= 20 cm: Meldung "OK" (Ausgabe)
4. Sonst: Meldung "Zu dünn" (Ausgabe)

Jetzt dasselbe in Python:

````python
# Algorithmus: Mindestdicke Betonplatte prüfen

dicke_cm = 18  # Eingabe: Dicke in cm

# Verarbeitung nach Regeln
if dicke_cm >= 20:
    ergebnis = "OK: Mindestdicke eingehalten"
else:
    ergebnis = f"Zu dünn! {dicke_cm} cm < 20 cm Minimum"

# Ausgabe
print(ergebnis)
````

Beachtet die Struktur: **Eingabe** (Variable `dicke_cm`), **Verarbeitung** (`if/else`), **Ausgabe** (`print`). Das ist bei jedem Algorithmus gleich.

### Algorithmus vs. Programm

Ein **Algorithmus** beschreibt abstrakt, wie ein Problem zu lösen ist. Ein **Programm** ist die konkrete Umsetzung in einer Sprache.

Derselbe Algorithmus ("Vergleiche zwei Zahlen") sieht in verschiedenen Sprachen unterschiedlich aus, die Logik bleibt aber identisch:

| Sprache | Code |
|---------|------|
| Deutsch | Wenn A grösser als B, dann gib A aus |
| Python  | `if a > b: print(a)` |
| BASIC   | `IF A > B THEN PRINT A` |

Ihr seht: Die **Logik** ist überall gleich, nur die **Syntax** (Schreibweise) unterscheidet sich.

### ✏️ Aufgabe 1: Euer erster eigener Algorithmus

Schreibt einen Algorithmus, der prüft, ob eine **Raumhöhe** die Mindestanforderung erfüllt.

Regeln:
- Wohnraum: Mindesthöhe 2.40 m
- Wenn die Höhe ausreicht: Meldung mit der Höhe und "OK"
- Wenn nicht: Meldung mit der Differenz zum Minimum

Testet euer Programm mit den Werten 2.70, 2.40 und 2.10.

````python
# Aufgabe 1: Raumhöhe prüfen

raumhoehe = 2.70       # Testet auch mit 2.40 und 2.10
mindesthoehe = 2.40

if raumhoehe >= mindesthoehe:
    print(f"✅ Raumhoehe {raumhoehe} m: OK (Minimum {mindesthoehe} m)")
else:
    differenz = mindesthoehe - raumhoehe
    print(f"❌ Raumhoehe {raumhoehe} m: Zu niedrig! Es fehlen {differenz:.2f} m")
````

### ✏️ Aufgabe 2: Algorithmus erweitern

Erweitert den Algorithmus: Die Mindesthöhe hängt jetzt von der **Nutzung** ab.

| Nutzung | Mindesthöhe |
|---------|-------------|
| Wohnraum | 2.40 m |
| Büro | 2.50 m |
| Lager | 2.20 m |

Tipp: Nutzt `if`, `elif` und `else`.

````python
# Aufgabe 2: Raumhöhe prüfen nach Nutzung

raumhoehe = 2.35
nutzung = "Wohnraum"   # Testet auch "Büro" und "Lager"

if nutzung == "Wohnraum":
    mindesthoehe = 2.40
elif nutzung == "Büro":
    mindesthoehe = 2.50
elif nutzung == "Lager":
    mindesthoehe = 2.20
else:
    mindesthoehe = 2.40
    print(f"⚠️ Nutzung '{nutzung}' unbekannt, verwende Standard 2.40 m")

if raumhoehe >= mindesthoehe:
    print(f"✅ {nutzung}: {raumhoehe} m >= {mindesthoehe} m")
else:
    differenz = mindesthoehe - raumhoehe
    print(f"❌ {nutzung}: {raumhoehe} m < {mindesthoehe} m (fehlen {differenz:.2f} m)")
````

### Vom BPMN-Diagramm zum Code

In der Vorlesung habt ihr einen Prozess (Innenbetonstütze) als Swimlane-Diagramm gezeichnet. Dieselbe Logik lässt sich in Python umsetzen.

Hier ein vereinfachtes Beispiel: Der Ablauf einer Stützenbemessung als Algorithmus.

````python
# Vereinfachter Bemessungsablauf als Algorithmus

# --- Eingaben (kommen normalerweise von der Architektin) ---
geschoss = "EG"
last_kn = 450        # Normalkraft in kN
breite_cm = 30       # Stützenbreite in cm
hoehe_cm = 30        # Stützenhöhe in cm

# --- Verarbeitung ---
flaeche_cm2 = breite_cm * hoehe_cm
flaeche_m2 = flaeche_cm2 / 10000
spannung_mpa = (last_kn / 1000) / flaeche_m2   # MN / m² = MPa

# Vereinfachte Prüfung: Betondruckfestigkeit C25/30 = 25 MPa
# Mit Sicherheitsfaktor: zulässig ca. 14 MPa
zulaessig_mpa = 14.0

# --- Ausgabe mit Bewertung ---
print(f"Stütze {geschoss}: {breite_cm}x{hoehe_cm} cm")
print(f"Normalkraft: {last_kn} kN")
print(f"Spannung: {spannung_mpa:.1f} MPa")
print(f"Zulässig: {zulaessig_mpa} MPa")
print()

if spannung_mpa <= zulaessig_mpa:
    print("✅ Nachweis erfüllt")
else:
    print("❌ Nachweis NICHT erfüllt")
    print(f"   Überschreitung: {spannung_mpa - zulaessig_mpa:.1f} MPa")
````

# Teil 2: Logische Verknüpfungen

In der Vorlesung habt ihr AND, OR und NOT kennengelernt. In Python heissen diese Operatoren `and`, `or` und `not` (kleingeschrieben).

| Logik | Python | Wahr wenn... |
|-------|--------|--------------|
| AND (∧) | `and` | **beide** Bedingungen wahr |
| OR (∨) | `or` | **mindestens eine** Bedingung wahr |
| NOT (¬) | `not` | Bedingung **umgekehrt** wird |

````python
# Logik im Bauingenieurwesen

tragfaehigkeit_ok = True
gebrauchstauglichkeit_ok = False

# AND: Beide müssen erfüllt sein
nachweis_erfuellt = tragfaehigkeit_ok and gebrauchstauglichkeit_ok
print(f"Tragfähigkeit OK: {tragfaehigkeit_ok}")
print(f"Gebrauchstauglichkeit OK: {gebrauchstauglichkeit_ok}")
print(f"Nachweis erfüllt (AND): {nachweis_erfuellt}")
print()

# OR: Mindestens eines muss zutreffen
riss_vorhanden = True
durchbiegung_zu_gross = False
sanierung_noetig = riss_vorhanden or durchbiegung_zu_gross
print(f"Riss vorhanden: {riss_vorhanden}")
print(f"Durchbiegung zu gross: {durchbiegung_zu_gross}")
print(f"Sanierung nötig (OR): {sanierung_noetig}")
print()

# NOT: Umkehrung
baubewilligung = False
baustopp = not baubewilligung
print(f"Baubewilligung erteilt: {baubewilligung}")
print(f"Baustopp (NOT): {baustopp}")
````

### Wahrheitstabellen mit Python

Statt die Tabellen von Hand auszufüllen, lassen wir Python die Arbeit machen. So seht ihr gleichzeitig, wie Schleifen funktionieren.

````python
# Wahrheitstabelle AND
print("A | B | A AND B")
print("--+---+--------")
for a in [0, 1]:
    for b in [0, 1]:
        ergebnis = a and b
        print(f"{a} | {b} |   {ergebnis}")
````

````python
# Wahrheitstabelle OR
print("A | B | A OR B")
print("--+---+-------")
for a in [0, 1]:
    for b in [0, 1]:
        ergebnis = a or b
        print(f"{a} | {b} |   {int(bool(ergebnis))}")
````

````python
# Wahrheitstabelle NOT
print("A | NOT A")
print("--+------")
for a in [0, 1]:
    ergebnis = int(not a)
    print(f"{a} |   {ergebnis}")
````

### ✏️ Aufgabe 4: Logik-Übungen aus der Vorlesung

Löst die folgenden Ausdrücke aus Slide 35. Berechnet sie zuerst **im Kopf** (Klammern zuerst!), dann prüft mit Python.

```
(1 AND 0) OR 1 = ?
NOT(1 AND 1) = ?
(0 OR 1) AND (1 OR 0) = ?
NOT(0 OR 1) AND 1 = ?
(1 OR 0) AND NOT(0 OR 1) = ?
(1 AND (0 OR 1)) OR (1 AND 0) = ?
(NOT 1 OR 0) AND (1 OR NOT 0) = ?
```

````python
# Aufgabe 4: Prüft eure Kopfrechnung

# Tragt zuerst eure Lösung ein, dann lasst Python prüfen
meine_loesung_1 = 1   # (1 AND 0) OR 1       → (0) OR 1 = 1
meine_loesung_2 = 0   # NOT(1 AND 1)         → NOT(1) = 0
meine_loesung_3 = 1   # (0 OR 1) AND (1 OR 0) → 1 AND 1 = 1
meine_loesung_4 = 0   # NOT(0 OR 1) AND 1    → NOT(1) AND 1 = 0 AND 1 = 0
meine_loesung_5 = 0   # (1 OR 0) AND NOT(0 OR 1) → 1 AND NOT(1) = 1 AND 0 = 0
meine_loesung_6 = 1   # (1 AND (0 OR 1)) OR (1 AND 0) → (1 AND 1) OR 0 = 1
meine_loesung_7 = 0   # (NOT 1 OR 0) AND (1 OR NOT 0) → (0 OR 0) AND (1 OR 1) = 0 AND 1 = 0

# Python berechnet die richtigen Antworten
python_1 = int(bool((1 and 0) or 1))
python_2 = int(not(1 and 1))
python_3 = int(bool((0 or 1) and (1 or 0)))
python_4 = int(bool(not(0 or 1) and 1))
python_5 = int(bool((1 or 0) and not(0 or 1)))
python_6 = int(bool((1 and (0 or 1)) or (1 and 0)))
python_7 = int(bool((not 1 or 0) and (1 or not 0)))

# Vergleich
aufgaben = [
    ("(1 AND 0) OR 1", meine_loesung_1, python_1),
    ("NOT(1 AND 1)", meine_loesung_2, python_2),
    ("(0 OR 1) AND (1 OR 0)", meine_loesung_3, python_3),
    ("NOT(0 OR 1) AND 1", meine_loesung_4, python_4),
    ("(1 OR 0) AND NOT(0 OR 1)", meine_loesung_5, python_5),
    ("(1 AND (0 OR 1)) OR (1 AND 0)", meine_loesung_6, python_6),
    ("(NOT 1 OR 0) AND (1 OR NOT 0)", meine_loesung_7, python_7),
]

print("Ergebnisse:")
print()
richtig = 0
for ausdruck, meine, korrekt in aufgaben:
    if meine == korrekt:
        print(f"  ✅ {ausdruck} = {korrekt}")
        richtig += 1
    else:
        print(f"  ❌ {ausdruck} = {korrekt} (ihr hattet: {meine})")

print(f"\n{richtig} von {len(aufgaben)} richtig")
````

# Teil 3: Binärsystem

Ein Computer kennt nur zwei Zustände: **0** und **1**. Aus diesen Bits wird alles aufgebaut: Zahlen, Text, Bilder, BIM-Modelle.

### Binär zu Dezimal

Jede Stelle hat einen Wert (von rechts: 1, 2, 4, 8, 16, 32, 64, 128). Man addiert nur dort, wo eine 1 steht.

````python
# Binär zu Dezimal: Schritt für Schritt

binaer = "00010101"

print(f"Binärzahl: {binaer}")
print()
print("Position:  7  6  5  4  3  2  1  0")
print(f"Bit:       {'  '.join(binaer)}")
print(f"Wert:    128 64 32 16  8  4  2  1")
print()

# Berechnung
summe = 0
for i, bit in enumerate(binaer):
    position = len(binaer) - 1 - i
    wert = int(bit) * (2 ** position)
    if int(bit) == 1:
        print(f"  Position {position}: 1 × 2^{position} = {wert}")
    summe += wert

print(f"\nSumme: {summe}")
print(f"Kontrolle mit Python: int('{binaer}', 2) = {int(binaer, 2)}")
````

### Dezimal zu Binär

Methode: Fortlaufend durch 2 teilen, Rest notieren, von unten nach oben lesen.

````python
# Dezimal zu Binär: Schritt für Schritt

zahl = 25

print(f"Dezimalzahl: {zahl}")
print()
print("Division  | Quotient | Rest")
print("----------+----------+-----")

reste = []
aktuell = zahl
while aktuell > 0:
    quotient = aktuell // 2
    rest = aktuell % 2
    print(f"{aktuell:>5} ÷ 2  |    {quotient:>3}   |  {rest}")
    reste.append(str(rest))
    aktuell = quotient

# Von unten nach oben lesen (umkehren)
binaer = ''.join(reversed(reste))
# Auf 8 Stellen auffüllen
binaer_8 = binaer.zfill(8)

print(f"\nErgebnis (von unten nach oben): {binaer}")
print(f"Auf 8 Bit aufgefüllt: {binaer_8}")
print(f"Kontrolle mit Python: bin({zahl}) = {bin(zahl)}")
````

### Grössenordnungen: Wie gross sind Daten?

Im Bauwesen arbeitet ihr mit grossen Datenmengen. Hier ein Gefühl für die Grössenordnungen:

````python
# Grössenordnungen im Bauwesen

beispiele = [
    ("Ein Buchstabe (ASCII)", 1, "Byte"),
    ("Eine Textzeile", 80, "Bytes"),
    ("Ein PDF-Plan (A3)", 5_000_000, "Bytes"),
    ("Ein Foto von der Baustelle", 8_000_000, "Bytes"),
    ("Ein kleines IFC-Modell", 50_000_000, "Bytes"),
    ("Ein grosses BIM-Modell", 500_000_000, "Bytes"),
    ("Punktwolke Laserscanning", 5_000_000_000, "Bytes"),
]

print(f"{'Beispiel':<35} {'Grösse':>15}")
print("-" * 52)

for name, groesse_bytes, _ in beispiele:
    if groesse_bytes < 1024:
        anzeige = f"{groesse_bytes} Byte"
    elif groesse_bytes < 1024**2:
        anzeige = f"{groesse_bytes/1024:.1f} kB"
    elif groesse_bytes < 1024**3:
        anzeige = f"{groesse_bytes/1024**2:.1f} MB"
    else:
        anzeige = f"{groesse_bytes/1024**3:.1f} GB"
    print(f"{name:<35} {anzeige:>15}")
````

# Teil 4: Daten, Information, Wissen

Die **Datenpyramide** zeigt, wie aus rohen Zeichen über mehrere Stufen Kompetenz entsteht:

```
  △ Kompetenz    → Erfahrungsbasierte Bewertung
 △△ Wissen       → Vernetzte Informationen
 △△△ Information  → Daten + Kontext
△△△△ Daten        → Zeichen + Syntax
△△△△△ Zeichen     → Einzelne Symbole
```

Im folgenden Beispiel seht ihr, wie Python-Variablen genau diese Stufen abbilden.

````python
# Die Datenpyramide in Python

# Stufe 1: Zeichen - einzelne Symbole, keine Bedeutung
zeichen = ['6', '.', '5', '0']
print(f"Zeichen:     {zeichen}")
print(f"  → Einzelne Symbole, keine Bedeutung")
print()

# Stufe 2: Daten - Zeichen zusammengesetzt nach Syntax
datum = 6.50
print(f"Daten:       {datum}")
print(f"  → Eine Zahl, aber was bedeutet sie? Preis? Länge? Gewicht?")
print()

# Stufe 3: Information - Daten + Kontext
info = {"Eigenschaft": "Geschossfläche", "Wert": 6.50, "Einheit": "m²"}
print(f"Information: {info['Eigenschaft']} = {info['Wert']} {info['Einheit']}")
print(f"  → Jetzt wissen wir, was die Zahl bedeutet")
print()

# Stufe 4: Wissen - Information + Einordnung
wissen = {
    "Eigenschaft": "Geschossfläche",
    "Wert": 6.50,
    "Einheit": "m²",
    "Norm": "SIA 416",
    "Kategorie": "Hauptnutzfläche"
}
print(f"Wissen:      {wissen['Eigenschaft']} nach {wissen['Norm']} = {wissen['Wert']} {wissen['Einheit']}")
print(f"             Kategorie: {wissen['Kategorie']}")
print(f"  → Eingeordnet in ein Regelwerk")
print()

# Stufe 5: Kompetenz - Erfahrungsbasierte Bewertung
print(f"Kompetenz:   'Eine Geschossfläche von {wissen['Wert']} {wissen['Einheit']}")
print(f"             für ein ganzes Gebäude? Das kann nicht stimmen.'")
print(f"  → Nur mit Erfahrung erkennbar, dass dieser Wert unrealistisch ist")
````

### Datentypen und die Datenpyramide

Letzte Woche habt ihr Python-Datentypen kennengelernt: `int`, `float`, `str`, `bool`. Diese Datentypen bestimmen, wie der Computer Daten speichert und verarbeitet. Hier seht ihr den Zusammenhang mit realen Bauwerks-Daten:

````python
# Datentypen in der Praxis: Bauteil-Informationen

bauteil = {
    "Name": "Stütze B2",               # String
    "Geschoss": 2,                       # Integer
    "Höhe_m": 3.50,                      # Float
    "Tragend": True,                     # Boolean
    "Material": "Beton C25/30",          # String
    "Normalkraft_kN": 450,               # Integer
    "Baujahr": "2025-03-15",             # String (ISO 8601)
}

print("Bauteil-Informationen:")
print()
for schluessel, wert in bauteil.items():
    typ = type(wert).__name__
    print(f"  {schluessel:<20} = {str(wert):<20} Typ: {typ}")

print()
print("Beachtet: Ohne die Schlüssel (Name, Geschoss...) wären die Werte")
print("nur Daten. Erst der Schlüssel macht sie zur Information.")
print("Das ist die Datenpyramide in Aktion!")
````

### ✏️ Aufgabe 9: Datenqualität prüfen

Hier ist ein Dictionary mit Informationen zu einer Betonplatte. Es enthält **absichtlich Fehler** in der Datenqualität.

Findet mindestens **3 Probleme** und benennt das jeweils verletzte DGIQ-Qualitätskriterium.

````python
# Aufgabe 9: Findet die Datenqualitätsprobleme

betonplatte = {
    "Name": "",                          # Problem?
    "Dicke_cm": 25,
    "Breite_m": 8.0,
    "Laenge_m": -5.0,                    # Problem?
    "Bewehrung": "ja",                   # Problem? (Vergleicht mit dem Typ oben)
    "Betonklasse": "C25/30",
    "Letzte_Aktualisierung": "2019-06-01", # Problem?
    "Geschoss": 2,
    "Einheit_Dicke": "cm",
    "Einheit_Breite": "m",
    "Einheit_Laenge": "feet",            # Problem?
}

print("Plattendaten:")
for k, v in betonplatte.items():
    print(f"  {k}: {v}")

print()
print("Welche Datenqualitätskriterien sind verletzt?")
print()

# Problem 1: Name ist leer ("")
# Kriterium: VOLLSTÄNDIGKEIT – Jedes Objekt muss identifizierbar sein.
print("Problem 1: Name ist leer")
print("  → Kriterium: Vollstaendigkeit")
print()

# Problem 2: Laenge_m = -5.0 (negative Laenge)
# Kriterium: FEHLERFREIHEIT – Eine negative Laenge ist physikalisch unmoeglich.
print("Problem 2: Laenge_m = -5.0 (negativ)")
print("  → Kriterium: Fehlerfreiheit")
print()

# Problem 3: Bewehrung = "ja" statt True/False
# Kriterium: EINHEITLICHKEIT / EINDEUTIGKEIT – Bool-Wert erwartet,
# String "ja" ist inkonsistent (was ist mit "Ja", "JA", "yes"?)
print("Problem 3: Bewehrung = 'ja' statt True")
print("  → Kriterium: Einheitlichkeit / Eindeutigkeit")
print()

# Problem 4: Letzte_Aktualisierung = "2019-06-01" (ueber 5 Jahre alt)
# Kriterium: AKTUALITÄT – Sind die Daten noch gueltig?
print("Problem 4: Letzte Aktualisierung 2019 (veraltet)")
print("  → Kriterium: Aktualitaet")
print()

# Problem 5: Einheit_Laenge = "feet" statt "m"
# Kriterium: EINHEITLICHKEIT – Breite in m, Dicke in cm, aber Laenge in feet?
# In der Schweiz wird metrisch gearbeitet.
print("Problem 5: Einheit_Laenge = 'feet' statt 'm'")
print("  → Kriterium: Einheitlichkeit")
````
