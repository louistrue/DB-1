## 🎯 Heureka – Fehler sichtbar machen

Du hast soeben drei Fehler automatisch gefunden.  
Jetzt zeigen wir sie im 3D-Modell – **grün = OK, rot = Fehler, orange = verwaist**.

Das ist ein Vorgeschmack auf Woche 9: Digitale Modellprüfung.

> **Eine einzige Lücke** – dann läuft das Ganze. Führe danach einfach aus und dreh das Modell.

````python
import plotly.graph_objects as go

# --- Fehler-IDs aus den Prüfungen sammeln ---

# Stäbe mit Profil-Fehler
staebe_fehler = set()
for msg in fehler_profil:
    # Format: 'FEHLER Profil | Stab HT-B2 | ...'
    stab_id = msg.split('|')[1].strip().replace('Stab ', '')
    staebe_fehler.add(stab_id)

# Knoten mit Geometrie-Fehler
knoten_geometrie_fehler = set()
for msg in fehler_geometrie:
    knoten_id = msg.split('|')[1].strip().replace('Knoten ', '')
    knoten_geometrie_fehler.add(knoten_id)

# Verwaiste Knoten
knoten_verwaist = set()
for msg in fehler_topologie:
    knoten_id = msg.split('|')[1].strip().replace('Knoten ', '')
    knoten_verwaist.add(knoten_id)

# Knoten-Dictionary für schnellen Zugriff
knoten_dict = {k['id']: k for k in modell['knoten']}

# --- Farben zuweisen ---
GRUEN  = '#2ECC71'
ROT    = '#E74C3C'
ORANGE = '#F39C12'
GRAU   = '#95A5A6'

def farbe_knoten(kid):
    if kid in knoten_geometrie_fehler: return ROT
    if kid in knoten_verwaist:         return ORANGE
    return GRUEN

def farbe_stab(sid):
    # AUFGABE: Gib ROT zurück wenn der Stab einen Fehler hat, sonst GRUEN.
    # HINWEIS: Prüfe ob sid in der Menge staebe_fehler enthalten ist.
    if ___:
        return ROT
    return GRUEN

# --- Plotly-Traces aufbauen ---
traces = []

# Stäbe als Linien (je Stab eine Trace mit 3 Punkten: Start, End, None für Lücke)
for typ, linienbreite, dash in [
    ('Haupttraeger', 5, 'solid'),
    ('Quertraeger',  3, 'solid'),
    ('Stuetze',      3, 'dot'),
]:
    for stab in modell['staebe']:
        if stab['typ'] != typ:
            continue
        s = knoten_dict[stab['start']]
        e = knoten_dict[stab['end']]
        farbe = farbe_stab(stab['id'])
        label = f"{stab['id']} | {stab['profil']}"
        if stab['id'] in staebe_fehler:
            label += '  ⚠ Falsches Profil!'
        traces.append(go.Scatter3d(
            x=[s['x'], e['x'], None],
            y=[s['y'], e['y'], None],
            z=[s['z'], e['z'], None],
            mode='lines',
            name=label,
            line=dict(color=farbe, width=linienbreite, dash=dash),
            hovertext=label,
            hoverinfo='text',
            showlegend=True,
        ))

# Knoten als Punkte
for knoten in modell['knoten']:
    farbe = farbe_knoten(knoten['id'])
    symbol = 'circle'
    label  = f"{knoten['id']} | {knoten['typ']} | z={knoten['z']}"
    if knoten['id'] in knoten_geometrie_fehler:
        label  += '  ⚠ z < 0!'
        symbol  = 'cross'
    elif knoten['id'] in knoten_verwaist:
        label  += '  ⚠ verwaist!'
        symbol  = 'diamond'
    traces.append(go.Scatter3d(
        x=[knoten['x']],
        y=[knoten['y']],
        z=[knoten['z']],
        mode='markers+text',
        name=label,
        marker=dict(size=8, color=farbe, symbol=symbol,
                    line=dict(color='white', width=1)),
        text=[knoten['id']],
        textposition='top center',
        textfont=dict(size=9, color=farbe),
        hovertext=label,
        hoverinfo='text',
        showlegend=False,
    ))

# Legende manuell
for farbe, name in [(GRUEN, 'OK'), (ROT, 'Fehler'), (ORANGE, 'Verwaist')]:
    traces.append(go.Scatter3d(
        x=[None], y=[None], z=[None],
        mode='markers',
        marker=dict(size=10, color=farbe),
        name=name,
        showlegend=True,
    ))

# --- Layout ---
n_fehler = len(alle_fehler)
fig = go.Figure(data=traces)
fig.update_layout(
    title=dict(
        text=f"Velobrücke Aaretal – Brücke 2 | Modellprüfung: {n_fehler} Fehler gefunden",
        font=dict(size=15)
    ),
    scene=dict(
        xaxis=dict(title='X (m)', backgroundcolor='#F8FAFC', gridcolor='#E2E8F0'),
        yaxis=dict(title='Y (m)', backgroundcolor='#F8FAFC', gridcolor='#E2E8F0'),
        zaxis=dict(title='Z (m)', backgroundcolor='#F8FAFC', gridcolor='#E2E8F0'),
        aspectmode='data',
        camera=dict(eye=dict(x=1.6, y=-1.6, z=0.8))
    ),
    paper_bgcolor='white',
    legend=dict(x=0.01, y=0.99, bgcolor='rgba(255,255,255,0.8)',
                bordercolor='#CBD5E1', borderwidth=1),
    margin=dict(l=0, r=0, t=50, b=0),
    height=600,
)

fig.show()
print()
print('Tipp: Modell drehen mit Maus | Hover über Element für Details | Legende klicken zum Ein-/Ausblenden')
````
