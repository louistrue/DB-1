## Prinzip

Mit einer `for`-Schleife geht Python jedes Element einer Liste nacheinander durch.
Das ist der einfachste Algorithmus: **Verarbeite jedes Element der Eingabe.**

```python

    # tue etwas mit element
```

````python
# Vollständiges Beispiel: alle Stäbe ausgeben
# Algorithmus: für jeden Stab -> Ausgabe einer Zeile

print("Alle Stäbe:")
print("-" * 52)

for stab in tragwerk['staebe']:
    print(f"  {stab['id']}  {stab['typ']:<12}  {stab['profil']}  {stab['laenge']:>5.1f} m")
````

# Teil 3: if-Bedingung – filtern und entscheiden

## Prinzip

Oft wollen wir nicht alle Elemente verarbeiten, sondern nur bestimmte.
Eine `if`-Bedingung ist eine **Verzweigung** im Algorithmus: ein Entscheidungspunkt.

```python
for element in liste:
    if bedingung:           # Entscheidung
        # nur wenn wahr
```

Vergleichsoperatoren:

| Operator | Bedeutung | Beispiel |
|----------|-----------|----------|
| `==` | gleich | `stab['typ'] == 'Stütze'` |
| `>` | grösser als | `stab['laenge'] > 10` |
| `<` | kleiner als | `stab['laenge'] < 5` |
| `and` | beide wahr | `typ == 'Träger' and laenge > 10` |

````python
# Vollständiges Beispiel: nur Stützen ausgeben
# Algorithmus: für jeden Stab -> wenn Typ = Stütze -> ausgeben

print("Nur Stützen:")
for stab in tragwerk['staebe']:
    if stab['typ'] == "Stütze":
        print(f"  {stab['id']}  {stab['profil']}  {stab['laenge']:.1f} m")
````

Wir wollen nicht nur ausgeben, sondern **berechnen**.
Das Muster: Variable auf 0 setzen, in der Schleife erhöhen.

```python

```

Das ist ein klassischer **Akkumulator-Algorithmus**: EVA in Reinform.
Eingabe: Liste. Verarbeitung: aufaddieren. Ausgabe: Summe.

# Teil 5: Funktionen – Algorithmen kapseln

## Prinzip

Bisher haben wir jeden Algorithmus einzeln geschrieben.
Eine **Funktion** gibt einem Algorithmus einen Namen und macht ihn wiederverwendbar:

```python
def funktionsname(parameter):
    # Algorithmus
    return ergebnis

# Aufrufen:
ergebnis = funktionsname(wert)
```

Das ist das EVA-Prinzip auf Funktionsebene:
- Eingabe = Parameter
- Verarbeitung = Code in der Funktion
- Ausgabe = `return`-Wert

````python
# Vollständiges Beispiel: Funktion gibt alle Stäbe eines Typs zurück

def get_staebe_by_typ(modell, typ):
    """Gibt alle Stäbe eines bestimmten Typs zurück.
    Eingabe:  modell (Dict), typ (String)
    Ausgabe:  Liste von Stäben
    """
    ergebnis = []
    for stab in modell['staebe']:
        if stab['typ'] == typ:
            ergebnis.append(stab)
    return ergebnis


# Einmal definiert, dreimal aufgerufen:
stuetzen    = get_staebe_by_typ(tragwerk, "Stütze")
traeger     = get_staebe_by_typ(tragwerk, "Träger")
quertraeger = get_staebe_by_typ(tragwerk, "Querträger")

print(f"Stützen:    {len(stuetzen)}")
print(f"Träger:     {len(traeger)}")
print(f"Querträger: {len(quertraeger)}")
````

# 🌟 Bonus: 3D-Visualisierung

Einfach ausführen. Das Modell wird als interaktives 3D-Diagramm dargestellt.

Das ist im Kleinen das, was BIM-Viewer tun: eine Datenstruktur einlesen und grafisch darstellen.
Die Datenstruktur ist dieselbe wie unsere – nur das Rendering ist komplexer.

````python
import plotly.graph_objects as go

knoten_dict = {k['id']: k for k in tragwerk['knoten']}
farben = {"Stütze": "#4A90D9", "Träger": "#E8A838", "Querträger": "#5BAD6F"}

fig = go.Figure()

for typ, farbe in farben.items():
    xv, yv, zv = [], [], []
    for stab in tragwerk['staebe']:
        if stab['typ'] == typ:
            kv = knoten_dict[stab['von']]
            kb = knoten_dict[stab['bis']]
            xv += [kv['x'], kb['x'], None]
            yv += [kv['y'], kb['y'], None]
            zv += [kv['z'], kb['z'], None]
    fig.add_trace(go.Scatter3d(
        x=xv, y=yv, z=zv, mode='lines',
        line=dict(color=farbe, width=6),
        name=typ, legendgroup=typ
    ))

fig.add_trace(go.Scatter3d(
    x=[k['x'] for k in tragwerk['knoten']],
    y=[k['y'] for k in tragwerk['knoten']],
    z=[k['z'] for k in tragwerk['knoten']],
    mode='markers+text',
    marker=dict(size=6, color='#333333'),
    text=[k['id'] for k in tragwerk['knoten']],
    textposition='top center',
    name='Knoten', legendgroup='Knoten'
))

fig.update_layout(
    title=tragwerk['name'],
    scene=dict(xaxis_title='X (m)', yaxis_title='Y (m)', zaxis_title='Z (m)', aspectmode='data'),
    height=600
)
fig.show()
````
