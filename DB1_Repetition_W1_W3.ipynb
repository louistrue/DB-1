# Bevor es losgeht: Fehler lesen lernen

Der grösste Unterschied zwischen Studierenden die vorankommen und denen die steckenbleiben: **die einen lesen die Fehlermeldung, die anderen nicht.**

Hier sind die 5 häufigsten Fehler. Führt jede Zelle aus und **lest die Meldung**.

### Fehler 1: `SyntaxError` – Doppelpunkt vergessen

````python
# Fuehrt diese Zelle aus. Was sagt Python?

x = 10
if x > 5
    print("gross")
````

Python sagt `SyntaxError: expected ':'`. Es fehlt der **Doppelpunkt** nach `if x > 5`.

Korrigiert den Code oben und führt die Zelle nochmal aus.

### Fehler 2: `IndentationError` – Einrückung fehlt

````python
# Fuehrt diese Zelle aus.

x = 10
if x > 5:
print("gross")
````

`IndentationError: expected an indented block`. Nach dem Doppelpunkt **muss** eingerückt werden (4 Leerzeichen).

Korrigiert es und führt nochmal aus.

### Fehler 3: `NameError` – Variable falsch geschrieben

````python
# Fuehrt diese Zelle aus.

spannweite = 12.0
print(f"Spannweite: {spanweite} m")
````

`NameError: name 'spanweite' is not defined`. Python findet `spanweite` nicht, weil die Variable `spannweite` heisst (doppeltes n). **Tippfehler sind der häufigste Grund für NameError.**

### Fehler 4: `TypeError` – Falscher Datentyp

````python
# Fuehrt diese Zelle aus.

laenge = "12"
flaeche = laenge * 5
print(flaeche)
````

Kein Fehler, aber das Ergebnis ist `"1212121212"` statt `60`. Weil `laenge` ein **String** ist (`"12"`), nicht eine Zahl (`12`). `"12" * 5` wiederholt den Text 5x.

**Datentypen sind wichtig.** Zahlen ohne Anführungszeichen, Text mit.

### Fehler 5: `IndexError` – Ausserhalb der Liste

````python
# Fuehrt diese Zelle aus.

lasten = [120, 85, 95]
print(lasten[3])
````

`IndexError: list index out of range`. Die Liste hat 3 Elemente, aber **der Index startet bei 0**. Also: `[0]`, `[1]`, `[2]`. Element `[3]` gibt es nicht.
